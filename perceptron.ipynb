```python
class Perceptron:
    def __init__(self,n_inputs, learning_rate=0.1):
        
        self.weights=[0.0]*n_inputs
        self.bias =0.0
        self.lr = learning_rate
    
    def predict(self, input):
        total = sum(w*x for w,x in (self.weights,input))
        total+=self.bias
        return 1 if total>=0 else 0
    
    def train(self , training_data,epochs=10000):
        for epoch in range(epochs):
            errors =0
            for input,target in training_data:
                print(input,target)
                prediction = self.predict(input)
                error = prediction-target
                if error!=0:
                    errors +=1
                    for i in range(len(self.weights)):
                        self.weights[i] -= self.lr*error*input[i]
                    self.bias -= self.lr*error
            if errors ==0:
                    print(f"Converged at epoch {epoch + 1}")
                    return
        print(f"Did not converge after {epochs} epochs")
                    
                
        
        
```

**solve XOR with two layer**

```python
def xor_network(x1,x2):
    or_neuron = Perceptron(2)
    
    or_neuron.weights = [1.0, 1.0]
    or_neuron.bias = -0.5

    nand_neuron = Perceptron(2)
    nand_neuron.weights = [-1.0, -1.0]
    nand_neuron.bias = 1.5

    and_neuron = Perceptron(2)
    and_neuron.weights = [1.0, 1.0]
    and_neuron.bias = -1.5

    hidden1 = or_neuron.predict([x1, x2])
    hidden2 = nand_neuron.predict([x1, x2])
    output = and_neuron.predict([hidden1, hidden2])
    return output
    
    
```

```python
print("\n=== XOR Gate (multi-layer network) ===")
for inputs, expected in xor_data:
    result = xor_network(inputs[0], inputs[1])
    print(f"  {inputs} -> {result} (expected {expected})")
```

```python
class TwoLayerNetwork:
    def __init__(self,learning_rate =0.1):
        self.hidden_weights = [[random.uniform(-1,1),random.uniform(-1,1)] for _ in range(2)]
        self.hidden_bias = [random.uniform(-1,1),random.uniform(-1,1)]
        self.output_weights = [random.uniform(-1,1),random.uniform(-1,1)]
        self.output_bias = random.uniform(-1,1)
        self.lr = learning_rate
    def sigmoid(self,x):
        return 1/(1+ math.exp(-x))
    
    def forward(self, inputs):
        self.inputs=inputs
        self.hidden_outputs =[]
        for i in range(2):
            z_hidden=sum(w*x for w,x in zip(self.hidden_weights[i],inputs)) +self.hidden_bias[i]
            self.hidden_outputs.append(self.sigmoid(z_hidden))
            
        z_out= sum(w*x for w,x in zip(self.output_weights,self.hidden_outputs))+self.output_bias 
        output = self.sigmoid(z_out)
        return output
    
    def train(self,training_data,epochs):
        
        for epoch in range(epochs):
            total_error=0
            for input,target in training_data:    
                output=self.forward(input)
                error=target-output
                total_error +=error**2
                d_output = error* output*(1-output)
                saved_w_outputs= self.output_weights[:]
                hidden_deltas =[]
                for i in range(2):
                    h= self.hidden_outputs[i]
                    hd = d_output * saved_w_outputs[i]*h*(1-h)
                    hidden_deltas.append(hd)
                
                for i in range(2):
                    self.output_weights[i]+=self.lr*self.hidden_outputs[i]*d_output
                self.output_bias+=self.lr*d_output
                
                for i in range(2):
                    for j in range(len(input)):
                        self.hidden_weights[i][j] += self.lr*hidden_deltas[i]*input[j]
                    self.hidden_bias[i]+=self.lr*hidden_deltas[i]
        
        
```

```python
net = TwoLayerNetwork(learning_rate=2.0)
net.train(xor_data, epochs=10000)
for inputs, expected in xor_data:
    result = net.forward(inputs)
    predicted = 1 if result >= 0.5 else 0
    print(f"  {inputs} -> {result:.4f} (rounded: {predicted}, expected {expected})")
```
